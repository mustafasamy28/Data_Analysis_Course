```python
import pandas as pd
```

```python
df = pd.read_csv("../datasets/surveys_3.csv", index_col=0)
df
```

#### Notice that the formula affected just the numeric variable and the Normalized Variable not affected

```python
normalized_df = (df - df.min()) / (df.max() - df.min())
normalized_df
```

##### in case of Standardization, here's the needed formula:
- (df - df.mean()) / df.std()   

- recentering the mean = 0 , std = 1

```python
Standardization_df =  (df - df.mean()) / (df.std())
Standardization_df
```
