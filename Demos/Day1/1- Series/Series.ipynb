### Series
Series is a one-dimensional labeled array capable of holding any data type (integers, strings, floating point numbers,
Python objects, etc.). The axis labels are collectively referred to as the index.

```python
import pandas as pd
import numpy as np
```

##### Create series from NumPy array
number of labels in 'index' must be the same as the number of elements in array

```python
my_simple_series = pd.Series(np.random.randn(5), index=['a', 'b', 'c','d','e'])
my_simple_series
```

```python

my_simple_series.index
```

```python

my_simple_series.values
```

- df.resample is a method in pandas that is used to resample time-series data in a DataFrame object. 
- Time-series data is data that is indexed by time, such as stock prices or weather data.

- The df.resample method groups the data in a DataFrame by a specified time interval and applies an   aggregation function, such as mean, sum, or count, to each group. 
- This can be used to perform operations such as downsampling (reducing the frequency of the data), upsampling (increasing the frequency of the data), and smoothing (reducing noise in the data).

- **The df.resample method requires a time-based index for the DataFrame, which can be created using the pd.to_datetime function or by setting the datetime column as the index using df.set_index('datetime').**

```python
# The method takes a string argument that specifies the time interval for resampling, 
# such as 1D for daily or 1W for weekly, 
# or a Timedelta object that specifies the length of the resampling interval.
# df_weekly = df.resample('1W').mean()

my_series.resample('3min').sum()
```

```python
my_series.resample('3min', label='right').sum()
```

### methods commonly used for analyzing and manipulating time series data.

 1. **pd.to_datetime()** : This method is used to convert date/time strings to pandas datetime objects, which can be used as an index for time series data.


```python
# Create a DataFrame with a date column as a string
df = pd.DataFrame({'date': ['2022-01-01', '2022-01-02', '2022-01-03'], 'value': [10, 20, 30]})
df.info()
```

```python
# Convert the date column to a pandas datetime object to use it late


df['date'] = pd.to_datetime(df['date'])
df.info()
```

 2. **df.set_index()** : This method sets a specified column as the index of a DataFrame, which is useful when the DataFrame contains time series data.


```python
# Create a DataFrame with a date column and a value column
df = pd.DataFrame({'date': ['2022-01-01', '2022-01-02', '2022-01-03'], 'value': [10, 20, 30]})
df
```

```python
df.set_index('date',inplace=True)
df
```

3. **df.resample()**: This method is used to resample time series data to a lower or higher frequency, such as downsampling hourly data to daily data.

```python
# Create a DataFrame with a daily datetime index and a value column
df = pd.DataFrame({'value': [10,20,30]} , index= pd.date_range(start='2022-01-01', end='2022-01-03', freq='D'))
df
```

```python
# Resample the data to a weekly frequency and calculate the mean
# df_weekly = df.resample('W').mean()
# df_weekly

df_weekly = df.resample('W').mean()
df_weekly

```

```python
# Sure! Based on the data you provided, the following is an explanation of the below results:

# value
# 2022-01-02	15.0
# This result is obtained by resampling the original data to a weekly frequency and taking the mean 
# of each week. Since there are two values in the first week (January 1st and January 2nd), 
# the mean of those two values is calculated, which is (10+20)/2 = 15.0. Therefore, 
# the value of 15.0 represents the average value of the first week of January 2022.

# value
# 2022-01-09	30.0
# This result is obtained by resampling the original data to a weekly frequency and taking the mean 
# of each week. Since there is only one value in the second week of January (January 3rd), 
# the mean of that value is simply the value itself, which is 30.0. Therefore, 
# the value of 30.0 represents the average value of the second week of January 2022.
```
