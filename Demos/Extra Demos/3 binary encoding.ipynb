```python
import pandas as pd
import category_encoders as ce
```

```python
df = pd.read_csv('../datasets/surveys.csv')
df.info()
```

```python
df
```

```python
binaryEncoder = ce.BinaryEncoder()
```

```python
binaryEncoder.fit_transform(df[['sex', 'species_id']])
```

```python
df = pd.concat([df, binaryEncoder.fit_transform(df[['sex', 'species_id']])], axis=1)
del df['sex']
del df['species_id']
df.info()
```
